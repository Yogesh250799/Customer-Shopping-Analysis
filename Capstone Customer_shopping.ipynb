```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()   ##To convert upper case to lower case
df.columns = df.columns.str.replace(' ','_')  ##To replace space from this '_'
#why i am doing is bcoz while performing in sql it easily run without any errors
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#Creating a new col as age_grp
labels = ['Young','Adult','Middle_age','Senior']
df['age_grp'] = pd.qcut(df['age'], q=4, labels = labels)
# what qcut does is its splits the ages into 4 equal grps based on the data distribution and assigns the labels we defined
```

```python
df[['age','age_grp']].head(10)
```

```python
mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(mapping)
#.map() looks at each value in the column and replaces it with the corresponding value from the dictionary.Example: If a row has "Weekly", it becomes 7.
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()  ##here we see both the columns r same so will drop one of the columns
```

```python
df = df.drop('promo_code_used',axis=1)
```
